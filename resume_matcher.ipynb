```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
resume = """
I am a Python developer with knowledge of machine learning,
natural language processing, SQL and data analysis.
I have worked with pandas, NumPy and scikit-learn.
"""
```

```python
print(resume)
```

```python
job_description = """
We are looking for a Python developer with experience in
machine learning, natural language processing, SQL and data analysis.
Knowledge of TensorFlow, deep learning and cloud computing is preferred.
"""
```

```python
print(job_description)
```

```python
documents = [resume, job_description]
```

```python
vectorizer = TfidfVectorizer()
```

```python
vectors = vectorizer.fit_transform(documents)
```

```python
print(vectorizer.get_feature_names_out())
```

```python
similarity = cosine_similarity(vectors[0], vectors[1])

print(similarity)
```

```python
match_percentage = similarity[0][0] * 100

print(f"Resume Match: {match_percentage:.2f}%")
```

```python
import re
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

def preprocess_text(text):
    text = text.lower()

    text = re.sub(r'[^a-zA-Z\s]', '', text)

    text = re.sub(r'\s+', ' ', text).strip()

    words = text.split()

    words = [word for word in words if word not in stop_words]

    return " ".join(words)
```

```python
text = "I am a Python developer and I know machine learning."

words = preprocess_text(text)

print(words)
```

```python
skills = [
    "python",
    "java",
    "c++",
    "sql",
    "machine learning",
    "deep learning",
    "natural language processing",
    "nlp",
    "tensorflow",
    "pytorch",
    "scikit learn",
    "pandas",
    "numpy",
    "aws",
    "docker",
    "git",
    "html",
    "css",
    "javascript"
]
```

```python
resume_skills = []

for skill in skills:
    if skill in clean_resume:
        resume_skills.append(skill)

print(resume_skills)
```

```python
job_skills = []

for skill in skills:
    if skill in clean_job:
        job_skills.append(skill)

print(job_skills)
```

```python
missing_skills = []

for skill in job_skills:
    if skill not in resume_skills:
        missing_skills.append(skill)

print(missing_skills)
```

```python
matched_skills = []

for skill in job_skills:
    if skill in resume_skills:
        matched_skills.append(skill)

print(matched_skills)
```

```python
skill_match = len(matched_skills) / len(job_skills) * 100

print(f"Skill Match: {skill_match:.2f}%")
```
